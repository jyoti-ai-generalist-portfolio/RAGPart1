```python
### Langchain + Typesense + Groq LLM + RAG Application


from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import Typesense
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_groq import ChatGroq





```

```python
import os
from dotenv import load_dotenv
load_dotenv()
import numpy as np


groq_api_key = os.getenv("GROQ_API_KEY")

loader = TextLoader("rag_test_ai_essay.txt")
documents = loader.load()
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap = 100)
chunks = text_splitter.split_documents(documents)
print (len(chunks))
# Extract the raw text from each Document chunk
texts = [doc.page_content for doc in chunks]

embeddings = HuggingFaceEmbeddings(show_progress=True)
# Generate the actual numerical vectors
vector_embeddings = embeddings.embed_documents(texts)




```

```python
import typesense
from langchain_community.vectorstores import Typesense

# 1. Initialize the official Typesense client directly
client = typesense.Client({
    "nodes": [{
        "host": "ldanjbht51pwz3yop-1.a2.typesense.net",
        "port": 443,
        "protocol": "https"
    }],
    "api_key": "09LCWqccKSJhhSPxAiAycp57eIoRyiAM",
    "connection_timeout_seconds": 120  # Keeps the connection stable during batch loops
})

# 2. Pass that client into LangChain's vector store wrapper
docsearch = Typesense(
    typesense_client=client,
    embedding=embeddings,
    typesense_collection_name='lang-chain'
)

# 3. Add documents in smaller manageable batches
batch_size = 200
for i in range(0, len(chunks), batch_size):
    chunk_batch = chunks[i : i + batch_size]
    docsearch.add_documents(chunk_batch)
    print(f"Successfully uploaded batch {i // batch_size + 1}")
```

```python
query = "How to stress test the retrieval layer"
found_docs = docsearch.similarity_search(query)
print(found_docs[0].page_content)


```
